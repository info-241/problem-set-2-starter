# Question 0
## What happens when pilgrims attend the Hajj pilgrimage to Mecca?

What happens when a diverse set of people are brought together toward a common purpose? Perhaps it brings diverse sets of people together in religious community who learn about each others' commonalities. Or, perhaps, the event highlights the differences between groups.  

If one were to simply speak with people who *choose* to go on a pilgrimage and compare views to those who *choose* not to go on that pilgrimage, there are **obvious** issues of selection bias. People who choose to go may be more open to new experiences, or more religious, or more conservative or ... when compared to people who choose not to go. 

In order to test the causal effect of attending the Hajj on views about others, you would need to conduct a randomized experiment. [Clingingsmith, Khwaja and Kremer (2009)](https://dash.harvard.edu/handle/1/3659699) did just that! 

In 2006, Pakistan conducted a lottery to allocate visas to attend the Hajj. The researchers conduct surveys among people who applied to this visa lottery, and talk to two different groups of people: 

1. One group is a set of people who applied for the Hajj visa and were successful; 
2. The other group is a set of people who applied for the Hajj visa and were not successful.

Because this was a randomly assigned lottery, these two groups of people are randomly assigned, and so meet our requirements of a valid experiment! 

The random assignment and data is collected in the following way (detailed in the paper): 

- Pakistani nationals apply for a chance to attend the Hajj at a domestic bank. During the time period of the study, 2006, Saudi officials agreed to grant 150,000 visas. 
- Of the 135,000 people who applied for a visa, 59% of those who applied were successful. 
- The remainder of the visas were granted under a different allocation method that was not randomly assigned, and so this experiment cannot provide causal evidence about other allocation mechanisms. 
- Among the 135,000 who apply, the authors conduct a random sample and survey about these respondents views about others.
